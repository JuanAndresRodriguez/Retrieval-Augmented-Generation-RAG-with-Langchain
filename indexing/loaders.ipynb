## Document Loading

Document loading in langchain is the process of loading data from various sources like files, databases, web pages, etc. The data is loaded into a Document object which is part of the langchain library. The Document object contains the text content of the data and metadata like the source URL, title, etc. The Document object is then used for further processing like text analysis, summarization, etc.

[https://python.langchain.com/docs/integrations/document_loaders](https://python.langchain.com/docs/integrations/document_loaders/)

#### Load a txt

```python
from langchain.document_loaders import TextLoader

loader = TextLoader("text.txt")

documents = loader.load()
```

```python
documents[0].page_content
```

#### Load a Website

```python
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader(["https://python.langchain.com/docs/integrations/document_loaders/web_base", "https://www.espn.com"])
documents = loader.load()
```

```python
documents[0]
```

```python
documents[0].page_content
```

```python
documents[1]
```

#### Load a CSV

```python
from langchain_community.document_loaders.csv_loader import CSVLoader
loader = CSVLoader(
    file_path="Air_Quality.csv",
    content_columns=["Name"] # the columns i select 
)

documents = loader.load()
# the documents are split into chunks

# one chunk per row
print(len(documents))

documents[0].page_content
```

```python
# load moby dick into a langchain doc
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("moby-dick.pdf")
documents = loader.load()

print("number of pages: ", len(documents))

# Print first 500 characters of first page
print("First 500 characters of first page:")
print(documents[7].page_content[:500])
```
