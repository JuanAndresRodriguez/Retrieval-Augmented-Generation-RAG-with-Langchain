```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="deepseek-r1:1.5b"
)

llm.invoke("Hey").content
```

```python
# ALL used for OpenAI API is reusable with Ollama using deepseek-r1:1.5b locally

from langchain.prompts import HumanMessagePromptTemplate, SystemMessagePromptTemplate
from langchain.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate([
    SystemMessagePromptTemplate.from_template("You are a {occupation} named {name}. Get into character and pretend to be this role. Answer questions accordingly"),
    HumanMessagePromptTemplate.from_template("{input}")
])

chain = prompt | llm

chain.invoke({
    "occupation": "CCP spy undercover",
    "name": "Lee Hun Wang",
    "input": "I heard you are a spy. Is that true?"
}).content
```
